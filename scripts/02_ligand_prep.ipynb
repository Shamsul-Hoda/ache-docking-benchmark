```python
from rdkit import Chem
from rdkit.Chem import AllChem

smiles = "COc1cc2c(cc1OC)C(=O)C(CC1CC[NH+](Cc3ccccc3)CC1)C2"
mol = Chem.MolFromSmiles(smiles)
mol = Chem.AddHs(mol)

print("Total atoms (with hydrogens):", mol.GetNumAtoms())
print("Formal charge:", Chem.GetFormalCharge(mol))
mol
```

```python
params = AllChem.ETKDGv3()
params.randomSeed = 42
embed_result = AllChem.EmbedMolecule(mol, params)
print("3D embedding (0 = success):", embed_result)

opt_result = AllChem.MMFFOptimizeMolecule(mol, maxIters=2000)
print("Minimisation (0 = converged):", opt_result)

writer = Chem.SDWriter("../prep/donepezil_3d.sdf")
writer.write(mol)
writer.close()
print("Saved: prep/donepezil_3d.sdf")
```
