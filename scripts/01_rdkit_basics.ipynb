```python
from rdkit import Chem

donepezil = Chem.MolFromSmiles("COc1cc2c(cc1OC)C(=O)C(CC1CCN(Cc3ccccc3)CC1)C2")
donepezil
```

```python
from rdkit.Chem import Descriptors, rdMolDescriptors

print("Formula:", rdMolDescriptors.CalcMolFormula(donepezil))
print("Molecular weight:", round(Descriptors.MolWt(donepezil), 2))
print("logP:", round(Descriptors.MolLogP(donepezil), 2))
print("H-bond donors:", Descriptors.NumHDonors(donepezil))
print("H-bond acceptors:", Descriptors.NumHAcceptors(donepezil))
print("Rotatable bonds:", Descriptors.NumRotatableBonds(donepezil))
print("Heavy atoms:", donepezil.GetNumHeavyAtoms())
```

```python
from rdkit.Chem import Draw

drugs = {
    "Donepezil": "COc1cc2c(cc1OC)C(=O)C(CC1CCN(Cc3ccccc3)CC1)C2",
    "Tacrine": "Nc1c2CCCCc2nc2ccccc12",
    "Rivastigmine": "CCN(C)C(=O)Oc1cccc(c1)[C@H](C)N(C)C",
    "Galantamine": "CN1CC[C@@]23C=C[C@@H](C[C@@H]2OC4=C(C=CC(=C34)C1)OC)O",
}

mols = []
for name, smiles in drugs.items():
    mol = Chem.MolFromSmiles(smiles)
    mols.append(mol)
    print(name, "| MW:", round(Descriptors.MolWt(mol), 1),
          "| logP:", round(Descriptors.MolLogP(mol), 2),
          "| HBD:", Descriptors.NumHDonors(mol),
          "| RotB:", Descriptors.NumRotatableBonds(mol))

Draw.MolsToGridImage(mols, legends=list(drugs.keys()), molsPerRow=4)
```
